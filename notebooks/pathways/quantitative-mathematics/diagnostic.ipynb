## Mathematics for Quantitative Studies diagnostic

Try each question before opening its solution.

### 1. Solve $2(x-3)+4=10$ over the real numbers.

<details><summary>Solution</summary>

Distribute to get $2x-6+4=10$, combine to $2x-2=10$, add 2, and divide by 2. Substitution gives $2(3)+4=10$ after evaluating $x-3=3$, so $x=6$ is verified.

</details>

### 2. For $f(x)=\sqrt{x-1}/(x-3)$ over the reals, select every required domain restriction.

- [ ] $x\ge 1$

- [ ] $x\ne3$

- [ ] $x>3$

- [ ] $x\ne1$

<details><summary>Solution</summary>

The radicand requires $x-1\ge0$, while the denominator excludes $x=3$. Thus the domain is $[1,3)\cup(3,\infty)$.

</details>

### 3. Compute $\det\begin{pmatrix}2&1\\3&4\end{pmatrix}$.

<details><summary>Solution</summary>

The determinant is $2(4)-1(3)=8-3=5$. Its nonzero value also certifies that the matrix is invertible.

</details>

### 4. What must be true if nonzero $v$ is an eigenvector of a matrix $A$?

- [ ] $Av$ is a scalar multiple of $v$.

- [ ] $Av=0$ for every eigenvector.

- [ ] $v$ must be a column of $A$.

- [ ] $A$ must be symmetric.

<details><summary>Solution</summary>

By definition, $Av=\lambda v$ for some scalar $\lambda$ and nonzero $v$. Symmetry is not required for the definition.

</details>

### 5. Differentiate $x^3-4x+2$ with respect to $x$.

<details><summary>Solution</summary>

Linearity and the power rule give $3x^2-4+0=3x^2-4$.

</details>

### 6. Evaluate $\lim_{x\to2}(x^2-4)/(x-2)$.

<details><summary>Solution</summary>

For $x\ne2$, the quotient equals $x+2$. The nearby values therefore approach $4$, even though the original quotient is undefined at 2.

</details>

### 7. Evaluate exactly $\int_0^1 3x^2\,dx$.

<details><summary>Solution</summary>

An antiderivative is $x^3$. The Fundamental Theorem gives $1^3-0^3=1$.

</details>

### 8. Which hypothesis is sufficient for the usual evaluation form of the Fundamental Theorem on $[a,b]$?

- [ ] $f$ is continuous on $[a,b]$.

- [ ] $f$ is positive on $[a,b]$.

- [ ] $f$ is a polynomial of degree at most two.

- [ ] $a$ and $b$ are integers.

<details><summary>Solution</summary>

Continuity on the closed interval is a standard sufficient hypothesis ensuring integrability and an antiderivative relationship suitable for evaluation.

</details>

### 9. For $P' = rP(1-P/K)$ with $r>0$ and $K>0$, select every equilibrium solution.

- [ ] $P(t)=0$

- [ ] $P(t)=K$

- [ ] $P(t)=r$

- [ ] $P(t)=Kt$

<details><summary>Solution</summary>

Set the right side to zero: $rP(1-P/K)=0$. Since $r>0$, either $P=0$ or $P=K$.

</details>

### 10. Which method directly applies to $y'=xy$ before imposing an initial condition?

- [ ] Separation of variables

- [ ] Gaussian elimination

- [ ] Integration by parts twice

- [ ] The quadratic formula

<details><summary>Solution</summary>

Writing $dy/y=x\,dx$ separates the variables on intervals where $y\ne0$; the equilibrium $y=0$ must also be retained.

</details>

### 11. Negate: “For every real $x$, there exists an integer $n$ with $n>x$.”

- [ ] There exists a real $x$ such that every integer $n$ satisfies $n\le x$.

- [ ] For every real $x$, every integer $n$ satisfies $n\le x$.

- [ ] There exists an integer $n$ such that every real $x$ satisfies $n\le x$.

- [ ] There exists a real $x$ and an integer $n$ with $n>x$.

<details><summary>Solution</summary>

The negation of $\forall x\,\exists n\,(n>x)$ is $\exists x\,\forall n\,(n\le x)$.

</details>

### 12. Select the indispensable components of a proof by ordinary induction for all integers $n\ge n_0$.

- [ ] Verify the base case $P(n_0)$.

- [ ] Prove $P(k)\Rightarrow P(k+1)$ for arbitrary $k\ge n_0$.

- [ ] Check the first one hundred cases.

- [ ] Assume $P(k+1)$ and derive $P(k)$.

<details><summary>Solution</summary>

The base establishes the chain and the induction step propagates truth from each $k$ to $k+1$. Checking many cases is evidence, not proof.

</details>

### 13. If $P(A\cap B)=0.18$ and $P(B)=0.30$, compute $P(A\mid B)$.

<details><summary>Solution</summary>

$P(A\mid B)=P(A\cap B)/P(B)=0.18/0.30=0.6$. The denominator is valid because $P(B)>0$.

</details>

### 14. A random variable equals 0 with probability 0.7 and 5 with probability 0.3. Find its expectation.

<details><summary>Solution</summary>

$E[X]=0(0.7)+5(0.3)=1.5$. This is a long-run average, not necessarily a possible single outcome.

</details>

### 15. Which statement correctly interprets a frequentist 95% confidence procedure?

- [ ] Across repeated samples, about 95% of intervals made by the procedure contain the fixed parameter.

- [ ] The fixed parameter has probability 0.95 of lying in this already computed interval.

- [ ] Exactly 95% of observations lie inside every confidence interval.

- [ ] The null hypothesis is true with probability 0.05.

<details><summary>Solution</summary>

In the frequentist model the parameter is fixed and intervals vary across samples. The 95% rate belongs to the procedure under its assumptions.

</details>

### 16. A $p$-value is best described as which probability?

- [ ] Under the null model, the probability of a result at least as incompatible with it as the observed result.

- [ ] The probability that the null hypothesis is true.

- [ ] The probability that the study will replicate.

- [ ] The size of the practical effect.

<details><summary>Solution</summary>

A $p$-value is a tail probability computed under the null model. It is neither the probability of the null nor an effect-size measure.

</details>

### 17. What does `sum(i*i for i in range(4))` return?

<details><summary>Solution</summary>

The generator produces $0^2,1^2,2^2,3^2$, whose sum is $0+1+4+9=14$.

</details>

### 18. Why can `0.1 + 0.2 == 0.3` be false in ordinary floating-point Python?

- [ ] Most decimal fractions are not represented exactly in finite binary floating point.

- [ ] Python randomizes decimal addition.

- [ ] Equality is undefined for numbers.

- [ ] Addition has lower precedence than equality.

<details><summary>Solution</summary>

Binary floating point stores nearby representable numbers, so arithmetic may contain a small rounding discrepancy. Compare with a justified tolerance when approximation is intended.

</details>
